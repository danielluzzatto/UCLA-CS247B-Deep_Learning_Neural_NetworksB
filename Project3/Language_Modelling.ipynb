### Observation and Analysis

Please answer the following questions. 

1. What can we say about the generated text in terms of grammar and coherence? 

Local word pairs look fluent ("Once upon a time", "she would", "play with"), but anything beyond ~2 words is incoherent: no consistent subject, plot, or tense, broken punctuation (.Once upon a worm), and characters that appear and vanish randomly.

2. What are the limitations of the Bigram language model?

The bigram is a 1st-order Markov model, so it only conditions on the previous token and has no memory of earlier context. That means it can't capture long-range dependencies like subject-verb agreement, coreference, or matching quotes, can't generalize semantically across tokens (it's effectively a vocab x vocab lookup), and easily falls into shallow repetition loops.

3. If the model is scaled with more parameters do you expect the bigram model to get substantially better? Why or why not?

Scaling won't help substantially. The bottleneck is the model class, not the parameter count: P(x_t | x_{t-1}) is essentially a lookup table that a small model already fits well, and adding parameters doesn't feed the model any more context. Real improvement requires an architecture that conditions on a longer history (n-gram, RNN, transformer), not a bigger bigram.

### Single Head Causal Attention (20 points)

We will first implement the single head causal attention layer. This layer is the same as the scaled dot product attention layer but with a causal mask to prevent the model from looking into the future.

Recall that Each head has a Key, Query and Value Matrix and the scaled dot product attention is calculated as : 

\begin{equation}
\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V
\end{equation}

where $d_k$ is the dimension of the key matrix.

Figure below from the original paper shows how the layer is to be implemented.

![image](./Images/Single_Head.png)

Image credits: [Attention is All You Need Paper](https://arxiv.org/abs/1706.03762)

### LayerNorm (10 points)

We will now implement the layer normalization layer. Layernorm is used across the model to normalize the activations of the previous layer. Recall that the equation for layernorm is given as:

\begin{equation}

\text{LayerNorm}(x) = \frac{x - \mu}{\sqrt{\sigma^2 + \epsilon}} \odot \gamma + \beta

\end{equation}

With the learnable parameters $\gamma$ and $\beta$. 

Remember that unlike batchnorm we compute statistics across the feature dimension and not the batch dimension, hence we do not need to keep track of running averages.

Please complete the `LayerNorm` class in `model.py`

```python
model = LayerNorm(MiniGPTConfig.embed_dim)
tests.check_layernorm(model, path_to_gpt_tester, device)
```

We implemented the encoder side of the transformer in model.py. 

We added an optional causal=True parameter to SingleHeadAttention and MultiHeadAttention. The default preserves the original decoder behavior so all the existing tests (check_singleheadattention, check_multiheadattention, check_transformer, check_miniGPT, etc.) continue to pass unchanged. When causal=False, the upper-triangular mask is simply not applied in forward, giving each token full bidirectional access to the rest of the sequence.

We also added a TransformerEncoderLayer class with the same pre-norm structure as TransformerLayer (LayerNorm -> attention -> residual -> LayerNorm -> feedforward -> residual), but instantiated with causal=False attention.

We also added a MiniEncoder class that mirrors MiniGPT: token embedding, learned positional embedding, embedding dropout, a stack of TransformerEncoderLayer blocks (count taken from config.num_layers), and a final LayerNorm. Unlike MiniGPT, the encoder does not have a language-modeling head or a generate method, since an encoder is meant to produce contextualized token representations rather than autoregressively predict the next token. It returns a tensor of shape (B, T, embed_dim) suitable for downstream tasks such as classification, retrieval, or as the encoder side of an encoder-decoder model.

Then we verified it, all seven original test cases still pass after the change. To confirm that the encoder is genuinely bidirectional, I ran the same input through both an encoder block and the original decoder block, then compared the representation of token 0 when the input was truncated to only the first three tokens versus the full five tokens. The decoder block produced an identical representation for token 0 in both cases (as expected, since the causal mask hides future tokens), while the encoder block produced a different representation (confirming that token 0 attends to tokens 1 through 4). This is the defining behavioral difference between an encoder and a causal decoder.
